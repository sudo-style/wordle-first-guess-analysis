```python
def wordle(guess, answer):
    guess = guess.lower()
    answer = answer.lower()
    result = ['⬜️'] * 5

    
    answer_letters = list(answer)
    
    # handle greens
    for i in range(5):
        if guess[i] == answer[i]:
            result[i] = '🟩'
            answer_letters[i] = None  # remove matched letter
    
    # handle yellows
    for i in range(5):
        if result[i] == '⬜️' and guess[i] in answer_letters:
            result[i] = '🟨'
            answer_letters[answer_letters.index(guess[i])] = None  # mark that letter as used
    
    return ''.join(result)
```

```python
# uses the same rules as the NYT wordle
wordle(guess = 'AAAAA', answer = 'AABAA')
```

```python
def mean_green(df):
    total_greens = 0
    total_entries = len(df)
    
    for output in df['Output']:
        total_greens += output.count('🟩')
    
    return total_greens / total_entries
```

This is the mean of how many green's are in the first guess
Ranging from 0 (if no greens were found) and 5 (if all guesses were perfect)
* $mean_{community} = 0.3381294964028777$
* $mean_{suspect} = 0.75$

Divide by 5 to have the range be from 0 to 1
* $\hat p_{community} = \frac{community_{mean}}{5} = 0.06762589928$
* $\hat p_{community} = \frac{community_{mean}}{5} = 0.15$

lengths of number of samples
* $n_{coummunity} = 139$
* $n_{suspect} = 56$

calculate the variance, multiplied by 5 because each sample has 5 letters
* $variance_{community}  = \frac{\hat p_{suspect} * (1 - \hat p_{suspect})}{(5 * c_{suspect})} = 0.00009072321874398417$
* $variance_{suspect}  = \frac{\hat p_{suspect} * (1 - \hat p_{suspect})}{(5 * c_{suspect})} = 0.0004553571428571429$

standard error difference
* $SE_{diff} = \sqrt{variance_{community} + variance_{suspect}} = 0.0233683624073474$

Z-score
* $z_score = \frac{\hat p_{community} - \hat p_{suspect}}{SE_{diff}} = 3.525026669970022$

```python
# the mean can range from [0 to 5]
# divide by 5 to have the range be from [0 to 1]
def p_mean_hat(mean, divisor = 5):
    output = mean / divisor
    return output
```

```python
def variance(n, mean, divisor = 5):
    p_hat_mean = p_mean_hat(mean, divisor)
    return p_hat_mean * (1 - p_hat_mean) / (n * divisor)
```

```python
def standard_error_difference(variance_1, variance_2):
    return (variance_1 + variance_2) ** 0.5
```

```python
def z_score(mean_1, mean_2, standard_error_difference):
    return (mean_1 - mean_2) / standard_error_difference
```

```python
from scipy.stats import norm
def p_value_right_tail(z_score):
    return 1 - norm.cdf(z_score)
```

```python
# number of first time guesses
n_community = len(community_subset)
n_suspect = len(subset_suspect)
```

```python
mean_green_community = mean_green(community_subset)
mean_green_suspect= mean_green(subset_suspect)
```

```python
# calculate p_mean_hat
p_mean_hat_community = p_mean_hat(mean_green_community)
p_mean_hat_suspect = p_mean_hat(mean_green_suspect)
```

```python
# calculate the variance
variance_community = variance(n_community, mean_green_community)
variance_suspect = variance(n_suspect, mean_green_suspect)
```

```python
# calculate the standard error difference
standard_error_difference_suspect = standard_error_difference(variance_community, variance_suspect)
standard_error_difference_suspect
```

```python
# calculate the z_score
z_score_suspect = z_score(p_mean_hat_suspect, p_mean_hat_community, standard_error_difference_suspect)
z_score_suspect
```

```python
# calculate the p_value
p_value_right_tail_suspect = p_value_right_tail(z_score_suspect)
p_value_right_tail_suspect
```

The null hypothesis is that the suspect is playing fairly

```python
def nullHypothesis(df_suspect, df_community, alpha = 0.01):
    print("Null hypothesis: The suspected user's mean number of green letters in the first guess, are equal to that of the average players.\n")
    
    n_suspect = len(df_suspect)
    n_community = len(df_community)
    mean_green_suspect = mean_green(df_suspect)
    mean_green_community = mean_green(df_community)
    print(f"Sample size (suspect): {n_suspect}")
    print(f"Sample size (community): {n_community}")
    print(f"Mean green letters (suspect): {mean_green_suspect:.3f}")
    print(f"Mean green letters (community): {mean_green_community:.3f}\n")
    
    
    p_hat_suspect = p_mean_hat(mean_green_suspect)
    p_hat_community = p_mean_hat(mean_green_community)
    variance_suspect = variance(n_suspect, mean_green_suspect)
    variance_community = variance(n_community, mean_green_community)
    se_diff = standard_error_difference(variance_community, variance_suspect)
    print(f"Estimated proportion (suspect): {p_hat_suspect:.4f}")
    print(f"Estimated proportion (community): {p_hat_community:.4f}")
    print(f"Variance (suspect): {variance_suspect:.6f}")
    print(f"Variance (community): {variance_community:.6f}")
    print(f"Standard error of difference: {se_diff:.6f}\n")

    z = z_score(p_hat_suspect, p_hat_community, se_diff)
    p_val = p_value_right_tail(z)
    print(f"Z-score: {z:.3f}")
    print(f"P-value (right-tailed): {p_val:.5f}")
    print(f"Significance level (α): {alpha}")
    
    if p_val > alpha:
        print("Fail to reject the null (no significant evidence of cheating).")
    else:
        print("Reject the null hypothesis (evidence suggests the user may be cheating).")
    
```

```python
# before game 1570
nullHypothesis(subset_suspect, community_subset)
```

```python
# after game 1570, excluding 1600
nullHypothesis(subset_suspect_pre_1570, community_subset)
```

```python
# both combined
nullHypothesis(subset_suspect_post_1570, community_subset)
```
